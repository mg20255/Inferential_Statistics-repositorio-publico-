```python
import matplotlib.pyplot as plt
import seaborn as sns

# Assuming the last executed cell's output is the binomial data
# We'll re-run the binomial generation for clarity and to have the data in a variable
binomial_data = np.random.binomial(10, 0.5, 50)

plt.figure(figsize=(10, 6))
sns.histplot(binomial_data, bins=range(0, 12), kde=False, color='skyblue', edgecolor='black')
plt.title('Histograma de la Distribución Binomial (n=10, p=0.5)')
plt.xlabel('Número de Éxitos')
plt.ylabel('Frecuencia')
plt.xticks(range(0, 11))
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
n = 10
p = 0.5

# Calcular la media teórica de la distribución binomial
mean_binomial = n * p

# Calcular la desviación estándar teórica de la distribución binomial
std_dev_binomial = np.sqrt(n * p * (1 - p))

# Estandarizar los datos binomiales
standardized_data = (binomial_data - mean_binomial) / std_dev_binomial

# Calcular la probabilidad de que las variables estandarizadas sean menores o iguales a 10
probability = np.mean(standardized_data <= 10)

print(f"Media teórica (μ): {mean_binomial:.2f}")
print(f"Desviación estándar teórica (σ): {std_dev_binomial:.2f}")
print(f"Probabilidad P(Z <= 10): {probability:.4f}")
```

```python
# ==============================================================================
# INTRODUCCIÓN A NUMPY 01
# Réplica completa de la clase: https://www.youtube.com/watch?v=UCWFaowRJgQ
# ==============================================================================

# ------------------------------------------------------------------------------
# 1. TIPOS DE OBJETOS Y ARITMÉTICA BÁSICA
# ------------------------------------------------------------------------------
print("=== 1. TIPOS DE OBJETOS Y ARITMÉTICA BÁSICA ===")
print(1 + 1)

# Conocer el tipo de objeto con la función type()
print("Tipo de 1 + 1:", type(1 + 1))             # int (entero)
print("Tipo de 1 + 1 + 0.5:", type(1 + 1 + 0.5)) # float (flotante)

# Declaración de variables y cadenas de caracteres (strings)
x = "Hola a todos"
print("Valor de x:", x)
print("Tipo de x:", type(x))                     # str (cadena)
print()

# ------------------------------------------------------------------------------
# 2. DEFINICIÓN Y PROPIEDADES DE LISTAS
# ------------------------------------------------------------------------------
print("=== 2. DEFINICIÓN Y PROPIEDADES DE LISTAS ===")
l1 = [1, 2, 3, 4, 5]
l2 = [6, 7, 8, 9, 10]

# Consultar el tamaño/longitud de una lista
print("Longitud de l1:", len(l1))

# Comparación booleana de longitudes
print("¿Tienen la misma longitud l1 y l2?:", len(l1) == len(l2))

# Concatenación de listas con el operador '+' (NO realiza suma vectorial)
print("Concatenación l1 + l2:", l1 + l2)
print()

# ------------------------------------------------------------------------------
# 3. ACCESO A ELEMENTOS DE UNA LISTA (INDEXACIÓN)
# ------------------------------------------------------------------------------
print("=== 3. ACCESO A ELEMENTOS DE UNA LISTA ===")
print("Primer elemento de l1 (índice 0):", l1[0])
print("Segundo elemento de l1 (índice 1):", l1[1])
print("Último elemento de l1 (índice 4):", l1[4])
# print(l1[5])  # Descomentar para ver el IndexError (fuera de rango)
print()

# ------------------------------------------------------------------------------
# 4. SUMA VECTORIAL MANUAL Y BUCLE FOR
# ------------------------------------------------------------------------------
print("=== 4. SUMA VECTORIAL MANUAL Y BUCLES FOR ===")
# Suma manual elemento a elemento
l_suma_manual = [l1[0]+l2[0], l1[1]+l2[1], l1[2]+l2[2], l1[3]+l2[3], l1[4]+l2[4]]
print("Suma manual elemento a elemento:", l_suma_manual)

# Uso de la función range() en un bucle for
print("\nIterando con range(0, 5):")
for i in range(0, 5):
    print("  Índice:", i)

# Suma elemento a elemento con comprensión de listas
l_suma_bucle = [l1[i] + l2[i] for i in range(len(l1))]
print("\nSuma usando bucle/comprensión de listas:", l_suma_bucle)

# Multiplicación por escalar usando comprensión de listas
m = [0.5 * l2[i] for i in range(len(l2))]
print("Multiplicación escalar por 0.5:", m)
print()

# ------------------------------------------------------------------------------
# 5. INTRODUCCIÓN A NUMPY Y ARREGLOS
# ------------------------------------------------------------------------------
print("=== 5. INTRODUCCIÓN A NUMPY Y ARREGLOS ===")
import numpy as np

# Convertir listas a arreglos de NumPy (np.ndarray)
l1_arreglo = np.array(l1)
l2_arreglo = np.array(l2)

print("Arreglo 1:", l1_arreglo)
print("Tipo de objeto:", type(l1_arreglo))

# Suma de vectores de forma directa
suma_vectorial = l1_arreglo + l2_arreglo
print("Suma vectorial directa con NumPy:", suma_vectorial)

# Multiplicación escalar directa
producto_escalar = 0.5 * l2_arreglo
print("Multiplicación por 0.5 directa con NumPy:", producto_escalar)
print()

# ------------------------------------------------------------------------------
# 6. CREACIÓN Y REDIMENSIONAMIENTO DE MATRICES (reshape)
# ------------------------------------------------------------------------------
print("=== 6. CREACIÓN Y REDIMENSIONAMIENTO DE MATRICES ===")
# Generar una lista de 20 elementos usando range(10, 30)
lista_20 = [i for i in range(10, 30)]
print("Lista original de 20 elementos:", lista_20)

# Convertir a arreglo NumPy
array_20 = np.array(lista_20)

# Redimensionar el vector a distintas matrices (filas, columnas)
matriz_5x4 = array_20.reshape(5, 4)
print("\nMatriz de 5 filas x 4 columnas:\n", matriz_5x4)

matriz_2x10 = array_20.reshape(2, 10)
print("\nMatriz de 2 filas x 10 columnas:\n", matriz_2x10)

matriz_10x2 = array_20.reshape(10, 2)
print("\nMatriz de 10 filas x 2 columnas:\n", matriz_10x2)
```
